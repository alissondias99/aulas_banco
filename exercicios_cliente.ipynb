```python
%%sql

create table CLIENTE (
    ID_cliente integer primary key AUTOINCREMENT,
    Nome varchar(155),
    Sexo check(Sexo in ("M", "F")),
    Email varchar(155),
    CPF varchar(11)
);

create table Endereco(
    Id_endereco integer primary key AUTOINCREMENT,
    Rua varchar(40),
    Bairro varchar(40),
    Cidade varchar(40),
    Estado char(2),
    IdCliente integer UNIQUE,
    foreign key (IdCliente) references Cliente(ID_cliente)
);

create table Telefone(
    Id_telefone integer primary key AUTOINCREMENT,
    Tipo check (Tipo in ("COM", "RES", "CEL")),
    Numero varchar(17),
    IdCliente integer,
    foreign key (IdCliente) references Cliente(ID_cliente)
);
```

```python
%%sql

-- Relatório geral dos clientes

select ID_cliente, Nome, Sexo, Email, CPF, Bairro, Rua, Cidade, Estado, Numero
from Cliente
    left join Endereco as E on ID_cliente == E.IDCliente
    left join Telefone as T on ID_cliente == T.IDCliente;
```

```python
%%sql 

--Relatório dos homens
select ID_cliente, Nome, Sexo, Email, CPF, Bairro, Rua, Cidade, Estado, Numero
from Cliente
    left join Endereco as E on ID_cliente == E.IDCliente
    left join Telefone as T on ID_cliente == T.IDCliente
where Sexo == "M";
-- alguns registros estão errados
```

```python
%%sql 

-- Relatório das mulheres
select ID_cliente, Nome, Sexo, Email, CPF, Bairro, Rua, Cidade, Estado, Numero
from Cliente
    left join Endereco as E on ID_cliente == E.IDCliente
    left join Telefone as T on ID_cliente == T.IDCliente
where Sexo == "F";
```

```python
%%sql

-- quantidades de homens e mulheres

SELECT Sexo, COUNT(Sexo) AS Total
FROM Cliente
GROUP BY Sexo;
```

```python
%%sql

--retorna só as mulheres que moram no centro de RJ e que não tenham telefone cadastrados

select ID_cliente, Nome, Email, Numero, Bairro, Cidade, Estado
from Cliente
    left join Endereco as E on ID_cliente == E.IDCliente
    left join Telefone as T on ID_cliente == T.IDCliente
where Sexo = "F" and Bairro = "CENTRO" and Estado = "RJ" and T.Numero is null;
```
