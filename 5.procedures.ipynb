```python
%%sql
CREATE DATABASE Cursos;
```

```python
%sql mysql+pymysql://root:@localhost/Cursos
```

```python
%%sql

CREATE Table cursos(
    ID_CURSO int PRIMARY key AUTO_INCREMENT,
    Nome VARCHAR(30),
    HORAS INT(30),
    PRECO FLOAT(10,2)
);
```

```python
%%sql

create PROCEDURE CRIAR_CURSO(P_NOME VARCHAR(30),
                             P_HORAS INT(30),
                             P_PRECO FLOAT(10,2))
BEGIN
    INSERT INTO cursos VALUES (NULL, P_NOME, P_HORAS, P_PRECO);
END
```

```python
%%sql

CALL CRIAR_CURSO("POWER BI", 17, 34.99);
CALL CRIAR_CURSO("Vire um programador em 3 meses", 8, 89.99);
CALL CRIAR_CURSO("Python", 41, 20.00);
```

```python
%%sql

--Criar procedure para consultar cursos

CREATE PROCEDURE ACHA_CURSO( P_IDCURSO INT )
BEGIN
        SELECT NOME, HORAS, PRECO FROM CURSOS
         WHERE ID_CURSO =  P_IDCURSO;
END
```

```python
%%sql

CALL ACHA_CURSO(3);
CALL ACHA_CURSO(2);
```
