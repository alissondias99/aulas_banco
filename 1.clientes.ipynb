```python
%%sql

CREATE TABLE cliente (
    Nome VARCHAR(100) NOT NULL,
    CPF VARCHAR(14) NOT NULL UNIQUE,
    Telefone VARCHAR(15),
    Email VARCHAR(100),
    Sexo CHAR(1),
    Endereco VARCHAR(255)
);

-- Está quebrando as regras da primeira FN
-- Campo Endereco deve ser quebrado em outra tabela com os campos Rua, número, Bairro, Cidade e Estado
-- Campos Telefone também deve ir para outra tabela pois ele deve ser multivalorado além de outras divisões.
```

```python
%%sql

-- Respeitando a 1º FN

create table Telefone(
    Id_telefone integer primary key AUTOINCREMENT,
    Tipo check (Tipo in ("COM", "RES", "TRAB")),
    Numero varchar(17),
    IdCliente integer,
    foreign key (IdCliente) references Cliente(ID_cliente)
);

create table Endereco(
    Id_endereco integer primary key AUTOINCREMENT,
    Bairro varchar(40),
    Rua varchar(40),
    Estado char(2),
    Cidade varchar(40),
    IdCliente integer UNIQUE,
    foreign key (IdCliente) references Cliente(ID_cliente)
);

create table Cliente(
    ID_cliente integer primary key AUTOINCREMENT,
    CPF VARCHAR(14) NOT NULL UNIQUE,
    Nome VARCHAR(100) NOT NULL,
    Email VARCHAR(100) UNIQUE,
    Sexo check (Sexo in ("F", "M"))
);
```
